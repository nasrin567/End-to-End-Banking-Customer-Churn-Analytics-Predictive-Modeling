import pandas as pd
import matplotlib.pyplot as plt

```python
results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Random Forest',
        'SVM Polynomial',
        'SVM RBF',
        'XGBoost'
    ],
    'Recall': [
        0.6520416711,
        0.6604278075,
        0.6711229947,
        0.7112229465,
        0.4304812834
    ],
    'Precision': [
        0.3283983849,
        0.4824218750,
        0.4960474308,
        0.4745153298,
        0.7488372093
    ],
    'F1': [
        0.4368845121,
        0.5575620767,
        0.5704545455,
        0.5689839572,
        0.5466893039
    ],
    'ROC_AUC': [
        0.7430787142,
        0.8306924246,
        0.8174833225,
        0.8302681689,
        0.8394883280
    ]
})

results
```

## Model Performance Comparison

```python
metrics = ['Recall', 'Precision', 'F1', 'ROC_AUC']

results.set_index('Model')[metrics].plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title('Model Performance Comparison')
plt.ylabel('Score')
plt.xlabel('Model')
plt.ylim(0, 1)
plt.xticks(rotation=20)
plt.legend(title='Metric')
plt.tight_layout()
plt.show()
```

## Best Model by Metric

```python
print("Best Recall:")
print(results.loc[results['Recall'].idxmax(), ['Model', 'Recall']])

print("\nBest Precision:")
print(results.loc[results['Precision'].idxmax(), ['Model', 'Precision']])

print("\nBest F1:")
print(results.loc[results['F1'].idxmax(), ['Model', 'F1']])

print("\nBest ROC-AUC:")
print(results.loc[results['ROC_AUC'].idxmax(), ['Model', 'ROC_AUC']])
```

## Overall Model Comparison

The models show different performance trade-offs across Recall, Precision, F1-score, and ROC-AUC.

SVM RBF achieves the highest Recall, making it effective at identifying customers who are likely to churn.

XGBoost achieves the highest Precision and ROC-AUC, indicating stronger overall discrimination and fewer false-positive churn predictions.

SVM Polynomial achieves the highest F1-score, although the difference from SVM RBF is small.

Logistic Regression performs weakest overall, particularly in Precision, F1-score, and ROC-AUC.

The final model should therefore be selected based on the business objective rather than a single evaluation metric.
